```python
import pandas as py
import numpy as np
import matplotlib.pyplot as plt
import yfinance as yf
import pandas_datareader
```

```python
apple = yf.download('AAPL', start= '2025-07-10', end='2026-07-10')
google = yf.download('GOOGL', start= '2025-07-10', end='2026-07-10')
```

```python
apple
```

```python
apple_close= apple['Close']
google_close= google['Close']
```

```python
google_close
```

```python
plt.figure(figsize=(22,4))
plt.plot(apple_close)
plt.plot(google_close)
plt.grid()
plt.show()
```

```python
weekly_apple_data= apple_close.resample('W').mean()
weekly_google_data= google_close.resample('W').mean()
```

```python

plt.plot(weekly_apple_data)
plt.plot(weekly_google_data)
plt.grid()
plt.show()
```

```python
plt.plot(weekly_apple_data, 'b--')
plt.plot(weekly_google_data, 'r--')
plt.grid()
plt.show()
```

```python
plt.plot(weekly_apple_data, 'b--', label='APPLE')
plt.plot(weekly_google_data, 'r--', label='GOOGLE')
plt.legend(loc='upper right', frameon=True)
plt.grid()
plt.title('weekly data', fontsize=14, weight='bold', color='green', pad=15)
plt.show()
```
